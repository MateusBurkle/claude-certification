# Create API Client

```python
from anthropic import Anthropic
client = Anthropic()
model = "claude-haiku-4-5-20251001"
```

# Make a request
## Now We are gonna add Multi-Tern Conversation
1. First delete message input typed down here
2. Create `function` called add_user_message(messages,text):
3. Create `function` called add_assistant_message(message,text):
4. Create `function` called chat(messages):

```python
def add_user_message(messages,text):
    user_message = {"role":"user", "content":text}
    messages.append(user_message)

def add_assistant_message(messages,text):
    assistant_message = {"role":"assistant","content":text}
    messages.append(assistant_message)

def chat(messages):
    message = client.messages.create(
        model = model,
        max_tokens= 1000,
        messages = messages
    )
    return message.content[0].text
```

```python
# Start with an empty message list
messages = []

# Add initial user question
add_user_message(messages,"Define quantum computing in one sentence")

# Get Claude's answer
answer = chat(messages)

# Add Claude's Response to the conversation history
add_assistant_message(messages,answer)

# Add User's follow up question 
add_user_message(messages,"Write other sentence")

# Get follow up response with full context 
final_answer = chat(messages)
final_answer
```
