````python
%pip install anthropic python-dotenv

from dotenv import load_dotenv
load_dotenv()

from anthropic import Anthropic
client = Anthropic()
model = "claude-haiku-4-5-20251001"



def add_user_message(messages, text):
    user_massage = {"role": "user", "content":text}
    messages.append(user_massage)

def add_assistant_message(messages, text):
    assistant_massage ={"role":"assistant", "content":text}
    messages.append(assistant_massage)

# Added stop_sequences at chat() and in params {} to work correctly
def chat(messages, system = None,temperature=1.0,stop_sequences=[]):
    params = {
        "model": model,
        "max_tokens":1000,
        "messages":messages,
        "temperature":temperature,
        # Line added 
        "stop_sequences":stop_sequences
    }
    if system:
        params["system"]= system

    message = client.messages.create(**params)
    return message.content[0].text
````

````python
# Doing exercise 

messages = []

prompt = """
Generate three different samples AWS CLI commands. Each should be very short.
"""
add_user_message(messages,prompt)

add_assistant_message(messages,"Here we got all three commands in a single block: \n```bash")
#Obs when are not using that message before the bash, the answer is different, 
#add_assistant_message(messages,"```bash")
text = chat(messages, stop_sequences=["```"])
text.strip()
````
